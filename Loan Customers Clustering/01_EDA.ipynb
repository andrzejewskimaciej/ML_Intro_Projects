Podział zmiennych ze względu na typy danych

```python
CATEGORICAL = ['address_state', 'home_ownership', 'loan_status', 'purpose', 'verification_status', 'emp_years', 'grade_numeric', 'sub_grade_numeric', 'emp_title']
NUMERICAL = ['annual_income',  'loan_amount', 'total_payment', 'installment', 'dti', 'int_rate', 'total_acc', 'emp_years', 'grade_numeric', 'sub_grade_numeric']
DATES = ['issue_date', 'last_credit_pull_date', 'last_payment_date', 'next_payment_date']
```

Boxploty zmiennych numerycznych

```python
plt.figure(figsize = (15, 7 * len(NUMERICAL)))
for i, col in enumerate(NUMERICAL):
    plt.subplot(len(NUMERICAL), 1, i+1)
    sns.boxplot(df[col])
    plt.title(f'{col}')
```

Histogramy zmiennych kategorycznych

```python
plt.figure(figsize=(20, (len(CATEGORICAL) - 1)* 7 ))

for index in range(len(CATEGORICAL)):
    if CATEGORICAL[index] == 'emp_title': continue
    plt.subplot(len(CATEGORICAL), 1, index + 1)

    sns.histplot(df[CATEGORICAL[index]])

    if df[CATEGORICAL[index]].unique().shape[0] > 5:
        plt.xticks(rotation=-45)
```

Histogramy i KDE zmiennych numerycznych

```python
plt.figure(figsize=(20, len(NUMERICAL)*7*2))

for index in range(len(NUMERICAL)):
    plt.subplot(len(NUMERICAL)*2, 1, (2*index + 1))

    column = NUMERICAL[index]
    if NUMERICAL[index] == 'annual_income':
        sns.histplot(df.loc[df.annual_income <df.annual_income.quantile(.95), 'annual_income'])
        plt.subplot(len(NUMERICAL)*2, 1, (2*index + 2))

        sns.kdeplot(df.loc[df.annual_income <df.annual_income.quantile(.95), 'annual_income'])
    else:
        sns.histplot(df[column])
        plt.subplot(len(NUMERICAL)*2, 1, (2*index + 2))
        sns.kdeplot(df[column])
```

Kolumny o wysokiej korelacji nie są zaskoczeniem, oczywiste że wysokość kredytu będzie skorelowana z całkowitą wpłatą oraz ratą, a także że procent kredytu będzie skorelowany z oceną kredytobiorcy (int_rate & grade, subgrade)

```python
sns.heatmap(df[NUMERICAL].corr(), annot=True)
```
